```python
class User:
    def __init__(self, username, name, email):
        self.username = username
        self.name = name
        self.email = email

    def __repr__(self):
        return f'User=(username={self.username}, name={self.name}, email={self.email})'

    def __str__(self):
        return self.__repr__
```

```python
virvi = User('virvihuta', 'Virvi Huta', 'hutavirvi07@gmail.com')
toni = User('tonhuta', 'Eltjon Huta', 'e.huta@mail.com')
reuf = User('reuf.ko', 'Reuf Kozlica', 'r.kozlica@mail.com')
elia = User('elia.h', 'Elia Huta', 'e.huta@mail.com')
aviel = User('aviel.h', 'Aviel Huta', 'a.huta@mail.com')
sui = User('sui.h', 'Sui Huta', 's.huta@gmail.com')
```
