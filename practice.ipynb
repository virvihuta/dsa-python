```python
def locate_cards(cards, query):
    low, high = 0, len(cards) - 1

    while low <= high:
        mid = (low + high) // 2
        mid_num = cards[mid]
        if mid_num == query:
            return mid
        elif mid_num > query:
            low = mid + 1
        elif mid_num < query:
            high = mid - 1

    return -1
```
